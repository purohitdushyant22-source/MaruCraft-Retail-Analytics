```python
df = df.rename(columns={
    'unit_price_inr': 'unit_price',
    'unit_cost_inr': 'unit_cost',
    'gross_item_sales_inr': 'sales',
    'discounted_item_revenue_inr': 'revenue',
    'item_cost_total_inr': 'total_cost',
    'estimated_item_gross_profit_inr': 'profit',
    'discount': 'discount_pct',
    'list_price_inr':'list_price',
    'shipping_cost_inr':'shipping_cost'
})
```

```python
df[['sales',
    'revenue',
    'total_cost',
    'profit']].head()
```

```python
df.columns.tolist()
```

```python
df['margin_pct'] = (
    df['profit']
    / df['revenue']
) * 100
```

```python
df['discount']=df['sales']-df['revenue']
```

```python
df['is_returned'] = (df['order_status'] == 'Returned').astype(int)
```

```python
df.head()
```

```python

df['month_name'] = df['order_date'].dt.strftime('%B')
df['quarter'] = df['order_date'].dt.quarter
```

```python
df['month_name_year'] = df['order_date'].dt.strftime('%b-%Y')
```

```python
import getpass
import os

import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

df = pd.read_csv("marucraft_cleaned.csv")


password = os.getenv("PGPASSWORD") or getpass.getpass(
    "PostgreSQL password for user 'postgres': "
)
if not password:
    raise ValueError("A PostgreSQL password is required to load the CSV.")

connection_url = URL.create(
    drivername="postgresql+psycopg2",
    username=os.getenv("PGUSER", "postgres"),
    password=password,
    host=os.getenv("PGHOST", "localhost"),
    port=int(os.getenv("PGPORT", "5432")),
    database=os.getenv("PGDATABASE", "marucraft_project"),
)
engine = create_engine(connection_url)

df.to_sql(
    "marucraft_orders",
    engine,
    if_exists="replace",
    index=False,
)
print(f"Loaded {len(df):,} rows into marucraft_project.marucraft_orders.")
```
